10.Write a python program to input mark in five subjects,calculate the percentage and display the corresponding grade.

```python
a=int(input("Enter the Math:"))
b=int(input("Enter the Physics:"))
c=int(input("Enter the Chemistry:"))
d=int(input("Enter the English:"))
e=int(input("Enter the Hindi:"))
x=a+b+c+d+e
t=500
p=(x/t)*100
print("Percentage=",p,"%")
if p>=90:
  print("Grade A")
elif 70<=p<90:
  print("Grade B")
elif 60<=p<70:
  print("Grade C")
elif 40<=p<60:
  print("Grade D")
else:
  print("Grade F")
```

Linear Regression model

```python
from sklearn.linear_model import LinearRegression
import numpy as np
x=np.array([2,3,5,6,7]).reshape(-1,1)
y=np.array([48,50,65,70,85])
m=LinearRegression()
m.fit(x,y)
a = m.intercept_
b = m.coef_[0]
print("intercept =",round(a,2))
print("slop =",round(b,2))
print("equation:")
print(f"marks = {round(a,2)}+{round(b,2)}*study hour")
h=10
predicted_marks = m.predict([[h]])
print("\npredicted marks =",round(predicted_marks[0],2))
if predicted_marks>=40:
  print("result:pass")
else:
  print("result:fail")

```

Ploting a graph of Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([2,3,5,6,7])
y=np.array([48,50,65,70,85])
mean_x=np.mean(x)
mean_y=np.mean(y)
numerator=np.sum((x-mean_x)*(y-mean_y))
denominator =np.sum((x-mean_x)**2)
slop = numerator/denominator
intercept=mean_y-slop*mean_x
y_predicted= intercept+slop*x

print("numerator :",numerator)
print("denominator :",denominator)
print("mean of x",mean_x)
print("mean of y",mean_y)
print("intercept =",intercept)

print("slop =",slop)
print("predicted value y=",y_predicted)
plt.figure(figsize=(10,5))
plt.scatter(x,y,color="green",label="data point")
plt.plot(x,y_predicted,color="blue",label=f"regration line(y={slop:.2f}x+{intercept:.2f})")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("simple linear regression")
plt.legend()
plt.grid(True)
plt.show()
```

Ice Cream(y)=[30,45,60,75,90]

```python
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
x =np.array([20,25,30,35,40]).reshape(-1,1)
y =np.array([30,45,60,75,90])
m=LinearRegression()
m.fit(x,y)
a=m.intercept_
b=m.coef_[0]
print("intercept=",round(a,2))
print("slope=",round(b,2))
print("Regression Equation:y={:2f}x+{:2f}".format(b,a))
temp_x=float(input("enter a new x value:"))
y_predicted_single = m.predict([[temp_x]])[0]
print("\n y_predicted =",round(y_predicted_single,2))

plt.scatter(x,y,color="blue",label="Data Point")
plt.plot(x,m.predict(x),color="red",label="Regression Line")
plt.scatter(temp_x,y_predicted_single,color="green",s=100,marker="*",label="predicted point")
plt.xlabel("x-axis:Temperature")
plt.ylabel("y-axis:Ice-cream Sale")
plt.title("Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```
