```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("/content/drive/MyDrive/ML LAB/house prediction.csv")
df.head()
```

```python
x=df[["sqft_living","bedrooms"]]
y=df.price
print(x,y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
m= LinearRegression()
m.fit(x_train,y_train)
m.predict(x_test)
print("intercept:",m.intercept_)
print("slope:",m.coef_)
```

```python
fig= plt.figure(figsize=(12,8))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(x["sqft_living"],x["bedrooms"],y)
ax.set_xlabel("sqft_living")
ax.set_ylabel("bedrooms")
ax.set_zlabel("price")
ax.set_title("Linear Regration")
```

```python
from sklearn.neighbors import KNeighborsClassifier
import seaborn as sns
from sklearn.metrics import accuracy_score
df=pd.read_csv("/content/drive/MyDrive/ML LAB/iris.csv")
df.head()
```

```python
x=df[["sepal_length","sepal_width"]]
y=df.species
print(x,y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
k=[3,5,7,9]
for i in k:
  m= KNeighborsClassifier(n_neighbors=i)
  m.fit(x_train,y_train)
  m.predict(x_test)
  print("accuracy:",m.score(x_test,y_test))
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x="sepal_length",y="sepal_width",hue="species",data=df)
plt.xlabel("sepal_length")
plt.ylabel("sepal_width")
plt.title("KNN")
plt.legend()
plt.show()
```

```python
x = df[["sepal_length", "sepal_width"]]
y = df.species
print(x)
print(y)
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
k = [3, 5, 7, 9]
accuracy_list = []
for i in k:
  m = KNeighborsClassifier(n_neighbors=i)
  m.fit(x_train, y_train)
  y_pred = m.predict(x_test)
  accuracy = m.score(x_test, y_test)
  accuracy_list.append(accuracy)
  print("K =", i)
  print("Accuracy =", accuracy)
  print()
```

```python
comparison = pd.DataFrame({"K Value": k,"Accuracy": accuracy_list})
print("KNN Comparison Table:")
print(comparison)

plt.figure(figsize=(8, 5))
plt.plot(k,accuracy_list,marker="o",linewidth=2,markersize=8)
plt.xlabel("K Value")
plt.ylabel("Accuracy")
plt.title("KNN Comparison for K = 3, 5, 7, 9")
plt.xticks(k)
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.bar([str(i) for i in k],accuracy_list)
plt.xlabel("K Value")
plt.ylabel("Accuracy")
plt.title("Accuracy Comparison of Different K Values")
plt.show()
best_accuracy = max(accuracy_list)
best_index = accuracy_list.index(best_accuracy)
best_k = k[best_index]
print("=================================")
print("Best K Value =", best_k)
print("Best Accuracy =", best_accuracy)
print("=================================")
```
