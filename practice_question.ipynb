10.Write a python program to input mark in five subjects,calculate the percentage and display the corresponding grade.

```python
a=int(input("Enter the Math:"))
b=int(input("Enter the Physics:"))
c=int(input("Enter the Chemistry:"))
d=int(input("Enter the English:"))
e=int(input("Enter the Hindi:"))
x=a+b+c+d+e
t=500
p=(x/t)*100
print("Percentage=",p,"%")
if p>=90:
  print("Grade A")
elif 70<=p<90:
  print("Grade B")
elif 60<=p<70:
  print("Grade C")
elif 40<=p<60:
  print("Grade D")
else:
  print("Grade F")
```
